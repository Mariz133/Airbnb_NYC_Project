#Milestone 3

###KMeans Clustering

```python
# Features for clustering
cluster_features = airbnb_model[
    ['price',
     'minimum_nights',
     'number_of_reviews',
     'availability_365',
     'latitude',
     'longitude',
     'room_type_Private_room',
     'room_type_Shared_room',
     'neighbourhood_group_Brooklyn',
     'neighbourhood_group_Manhattan',
     'neighbourhood_group_Staten_Island']
]
```

```python
scaler = MinMaxScaler()
scaled_features = scaler.fit_transform(cluster_features)
```

```python
inertia = []

k_values = range(1, 11)

for k in k_values:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(scaled_features)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(8,5))

plt.plot(k_values, inertia, marker='o')

plt.title("Elbow Method For Optimal K")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Inertia")

plt.show()
```

```python
# Create KMeans model
kmeans = KMeans(n_clusters=4, random_state=42)

# Fit model
kmeans.fit(scaled_features)

# Add cluster labels
airbnb_model['cluster'] = kmeans.labels_
```

##TVD Hypothesis Testing

```python
# Null Hypothesis:
# The distribution of Airbnb room types in Manhattan
# is the same as the distribution of room types in NYC overall.

# Alternative Hypothesis:
# The distribution of Airbnb room types in Manhattan
# is different from the distribution in NYC overall.
```

```python
nyc_prop = airbnb_NYC['room_type'].value_counts(normalize=True)

manhattan = airbnb_NYC[
    airbnb_NYC['neighbourhood_group'] == 'Manhattan'
]

manhattan_prop = manhattan['room_type'].value_counts(normalize=True)

prop_df = pd.DataFrame({
    'NYC': nyc_prop,
    'Manhattan': manhattan_prop
})

prop_df = prop_df.fillna(0)

prop_df
```

```python
tvd = np.abs(prop_df['NYC'] - prop_df['Manhattan']).sum()/2
tvd
```

```python
n = len(manhattan)

tvd_list = []

for i in range(1000):

    sample = airbnb_NYC.sample(n)

    sample_prop = sample['room_type'].value_counts(normalize=True)

    sim_df = pd.DataFrame({
        'NYC': nyc_prop,
        'Sample': sample_prop
    })

    sim_df = sim_df.fillna(0)

    sim_tvd = np.abs(sim_df['NYC'] - sim_df['Sample']).sum()/2

    tvd_list.append(sim_tvd)
```

```python
plt.figure(figsize=(8,5))

plt.hist(tvd_list)

plt.axvline(tvd, color='red')

plt.xlabel("TVD")
plt.ylabel("Frequency")

plt.title("Distribution of TVD Under Null Hypothesis")

plt.show()
```

```python
# Conclusion:
# The observed TVD is much larger than most simulated TVDs,
# suggesting that Manhattan has a different room type
# distribution compared to NYC overall.
```
